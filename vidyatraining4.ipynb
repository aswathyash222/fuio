**reverse a string**

reverse number

```python
n=int(input("enter a number:"))
op=0
while n>0:
  rem=n%10
  op=op*10+rem
  n=n//10
print("reversed number:",op)
```

**sum of** **digit**

```python
n=int(input("enter a number:"))
sum=0
while n>0:
  r=n%10
  sum=sum+r
  n=n//10
print("sum of the digit:",sum)
```

**largest digit**

```python
n=int(input("enter a number:"))
large=0
while n>0:
  r=n%10
  if r>large:
    large=r
  n=n//10
print("largest digit:",large)
```

**pallindrome or not**

```python
n=int(input("enter a number:"))
a=n
op=0
while(n>0):
  r=n%10
  op=op*10+r
  n=n//10
if (a==op):
    print("palindrome")
else:
    print("not a palindrome")
```

**string palindrome**

```python
n=(input("enter a string:"))
op=n[::-1]
if (n==op):
    print("palindrome")
else:
    print("not a palindrome")
```

**anagram or not**

```python
a=input("enter the first string:")
b=input("enter the second string:")
if sorted(a)==sorted(b):
  print("anagram")
else:
  print("not anagram")
```

**white space**

```python
n=input("enter a string:")
op=""
for ch in n:
  if ch!=" ":
    op=op+ch
print("string without space:",op)
```

**vowels or consonants count**

```python
n=input("enter a string:")
vowels=0
cons=0
for i in n:
  if(i in "aeiouAEIOU"):
    vowels+=1
  else:
    cons+=1
print("No. of vowels:",vowels)
print("No of consonants:",cons)
```
